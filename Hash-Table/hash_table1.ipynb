Two Sum

```python
class Solution(object):
    def twoSum(self, nums, target):
        seen = {}
        for i, num in enumerate(nums):
            comp = target-num
            if comp in seen:
                return [seen[comp], i]
            seen[num] = i
        return []
```

Longest Substring Without Repeating Characters

```python
class Solution(object):
    def lengthOfLongestSubstring(self, s):
        chr_index = {}
        left = 0
        max_length = 0
        for right in range(len(s)):
            if s[right] in chr_index:
                left = max(left,chr_index[s[right]]+1)
            chr_index[s[right]] = right
            max_length = max(max_length,right - left + 1)
            if max_length >= len(s) - left:
                break
        return max_length
        
```

Construct Binary Tree from Preorder and Inorder Traversal

```python
from typing import List, Optional

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def buildTree(self, preorder: List[int], inorder: List[int]) -> Optional[TreeNode]:
        inorder_index_map = {val: idx for idx, val in enumerate(inorder)}
        preorder_idx = 0

        def helper(in_left: int, in_right: int) -> Optional[TreeNode]:
            nonlocal preorder_idx
            
            if in_left > in_right:
                return None

            root_val = preorder[preorder_idx]
            preorder_idx += 1
            root = TreeNode(root_val)

            pivot = inorder_index_map[root_val]

      
            root.left = helper(in_left, pivot - 1)
            root.right = helper(pivot + 1, in_right)

            return root

        return helper(0, len(inorder) - 1)
```

Integer to Roman

```python
class Solution(object):
    def intToRoman(self, num):
        """
        :type num: int
        :rtype: str
        """
        res=[]
        thousands=num//1000
        if thousands>0:
            res.append('M'*thousands)
            num%=1000
        hundreds=num//100
        if hundreds>0:
            if hundreds==9:
                res.append('CM')
            elif hundreds>=5:
                res.append('D'+'C'*(hundreds-5))
            elif hundreds==4:
                res.append('CD')
            else:
                res.append('C'*hundreds)
            num%=100
        tens=num//10
        if tens>0:
            if tens==9:
                res.append('XC')
            elif tens>=5:
                res.append('L'+'X'*(tens-5))
            elif tens == 4:
                res.append('XL')
            else:
                res.append('X'*tens)
            num%=10
        ones=num
        if ones>0:
            if ones==9:
                res.append('IX')
            elif ones >= 5:
                res.append('V'+'I'*(ones-5))
            elif ones==4:
                res.append('IV')
            else:
                res.append('I'*ones)
        return ''.join(res)
```

Roman to Integer

```python
class Solution(object):
    def romanToInt(self, s):
        total = 0
        prev = 0
        for i in reversed(s):
            if i == 'I':
                val = 1
            elif i == 'V':
                val = 5
            elif i == 'X':
                val = 10
            elif i == 'L':
                val = 50
            elif i == 'C':
                val = 100
            elif i == 'D':
                val = 500
            elif i == 'M':
                val = 1000
            total += -val if val < prev else val
            prev = val
        return total
```

Substring with Concatenation of All Words

```python
class Solution(object):
    def findSubstring(self, s, words):
        if not s or not words:
            return []
        word_len = len(words[0])
        num_words = len(words)
        total_len = word_len * num_words
        s_len = len(s)
        word_count = Counter(words)
        result = []
        for i in range(word_len):
            left = i
            right = i
            seen = Counter()
            words_in_window =0
            while right+word_len <= s_len:
                current_word = s[right:right + word_len]
                right += word_len
                if current_word in word_count:
                    seen[current_word] += 1
                    words_in_window += 1
                    while seen[current_word] > word_count[current_word]:
                        left_word = s[left:left + word_len]
                        seen[left_word] -= 1
                        words_in_window -= 1
                        left += word_len
                    if words_in_window == num_words:
                        result.append(left)
                else:
                    seen.clear()
                    words_in_window = 0
                    left = right
        return result
```
