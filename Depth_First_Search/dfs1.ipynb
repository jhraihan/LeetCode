Flatten Nested List Iterator

```python
class NestedIterator(object):

    def __init__(self, nestedList):
        self.flat = []
        self.index = 0
        self._flat(nestedList)
    def _flat(self,nestedList):
        for item in nestedList:
            if item.isInteger():
                self.flat.append(item.getInteger())
            else:
                self._flat(item.getList())
        

    def next(self):

        res = self.flat[self.index]
        self.index += 1
        return res
        

    def hasNext(self):

        return self.index< len(self.flat)
```
