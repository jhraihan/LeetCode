Longest Substring Without Repeating Characters

```python
class Solution(object):
    def lengthOfLongestSubstring(self, s):
        
        chr_index = {}
        left = 0
        max_length = 0
        for right in range(len(s)):
            if s[right] in chr_index:
                left = max(left,chr_index[s[right]]+1)
            chr_index[s[right]] = right
            max_length = max(max_length,right - left + 1)
            if max_length >= len(s) - left:
                break
        return max_length
```

Longest Palindromic Substring

```python
class Solution(object):
    def longestPalindrome(self, s):
        res = ""
        for i in range(len(s)):
            temp = self.helper(s,i,i)
            if len(temp) > len(res):
                res= temp
            temp = self.helper(s,i,i+1)
            if len(temp) > len(res):
                res = temp
        return res
    def helper(self,s,l,r):
        while l>=0 and r<len(s) and s[l] == s[r]:
            l-=1;r+=1
        return s[l+1:r]
```

Zigzag Conversion

```python
class Solution(object):
    def convert(self, s, numRows):
        if numRows==1:
            return s
        result=[]
        cycle=2*numRows-2
        for i in range(numRows):
            for j in range(i,len(s),cycle):
                result.append(s[j])
                if i!=0 and i!=numRows-1 and j+cycle-2*i<len(s):
                    result.append(s[j+cycle-2*i])
        return ''.join(result)
```
